```python
import pandas as pd
```

```python
# 24/25 games dataframe
df = pd.read_csv('WFL2425games.csv')

df.head()
```

```python
# Expand each game into one row per total goal
# Result: columns ["GameID", "Goal"] where Goal counts from 1..(Team1_goals+Team2_goals)

# Ensure numeric goals (handle any missing as 0)
goals_df = df[['GameID', 'Team1_goals', 'Team2_goals']].copy()
goals_df['Team1_goals'] = goals_df['Team1_goals'].fillna(0).astype(int)
goals_df['Team2_goals'] = goals_df['Team2_goals'].fillna(0).astype(int)

goals_df['TotalGoals'] = goals_df['Team1_goals'] + goals_df['Team2_goals'] # create new column for total goals

# Filter games with at least one goal
nonzero_goals = goals_df[goals_df['TotalGoals'] > 0].copy()

# Repeat each game row TotalGoals times and enumerate per game
expanded = nonzero_goals.loc[nonzero_goals.index.repeat(nonzero_goals['TotalGoals'])].copy() # copying gameID by multiplying to number of total goals
expanded['Goal'] = expanded.groupby(level=0).cumcount() + 1 # adding goals up by 1

# Final expanded dataframe
expanded_goals_df = expanded[['GameID', 'Goal']].reset_index(drop=True)

# Preview
expanded_goals_df.head(30)
```
